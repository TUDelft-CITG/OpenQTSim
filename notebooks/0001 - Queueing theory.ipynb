# Queueing theory

Queueing theory describes systems in which customers arrive, wait when capacity is unavailable, receive service and leave. A customer can be a vessel, truck, container or person. In a port application, servers might be berths or cranes.

This introduction defines the quantities used throughout the book. Continue with [Simulation](0002%20-%20Simulation.ipynb) to see how individual events generate those quantities.

## Kendall notation

The shorthand **A/S/c** identifies the interarrival-time distribution, service-time distribution and number of parallel servers:

| Symbol | Meaning | Squared coefficient of variation |
|---|---|---|
| D | Deterministic: a fixed duration | 0 |
| M | Exponential durations; for arrivals, a Poisson process | 1 |
| E2 | Erlang distribution with shape 2 | 1/2 |

For example, M/E2/4 has exponential interarrival times, Erlang-2 service times and four servers. A/S/c alone does not specify queue discipline, finite capacity, customer priorities or abandonment. Unless stated otherwise, examples use first-come-first-served waiting and no abandonment. Shared-quay examples explicitly add vessel lengths and a quay-length constraint.

## Rates, workload and stability

Let $\lambda$ be the arrival rate, $\mu$ the service rate **per server**, and $c$ the number of identical servers. Mean interarrival time is $1/\lambda$ and mean service time is $1/\mu$. For a standard work-conserving queue, utilisation is

$$\rho = \frac{\lambda}{c\mu}.$$

A conventional stochastic queue with unlimited waiting room needs $\rho < 1$ for a steady-state distribution. As utilisation approaches one, spare capacity to absorb random peaks disappears. A deterministic queue can behave differently at exact balance; finite queues and abandonment also change the stability interpretation.

**API convention:** the analytical notation here uses per-server $\mu$. The parallel helper `openqtsim.mt_engine.create_tasks` instead takes **total** capacity as its `mu` argument; its per-server rate is `mu / c`. The example notebooks make that conversion explicitly.

## Performance measures

* $W_q$: mean waiting time before service.
* $W_s$: mean total time in the system, including service.
* $L_q$: mean number waiting.
* $L_s$: mean number in the system, including customers in service.

With consistent populations, observation conventions and a stable long-run regime, **Little's law** gives $L_q=\lambda W_q$ and $L_s=\lambda W_s$. Also, $W_s=W_q+E[ST]$ when every counted customer receives service.

Port service targets often use **$E[W_q]/E[ST]$**, the ratio of mean waiting time to mean service time. A 10% target is not a guarantee for each vessel, nor the probability that a vessel has to wait. Time averages over a finite simulation horizon can differ from averages over arriving customers; the basic examples show both.

## A reference solution: M/M/1

For an M/M/1 queue with independent arrivals and service, unlimited capacity, first-come-first-served service and $\lambda<\mu$:

$$W_q=\frac{\rho}{\mu-\lambda},\qquad W_s=\frac{1}{\mu-\lambda},\qquad L_q=\frac{\rho^2}{1-\rho}.$$

At $\lambda=8$ customers/hour and $\mu=9$ customers/hour, $\rho=8/9$, $W_q=8/9$ hours and $W_s=1$ hour. A finite random simulation will fluctuate around these long-run values, not reproduce them exactly.

The later M/M/1 examples compare simulation against this reference. M/M/c uses the Erlang-C solution. Erlang-2 and general distributions require other theory, approximations or simulation; M/M/c results should not be treated as exact for those systems.
